# Exploratory Data Analysis (EDA) of a Movie Dataset

```python
# Import libraries
import pandas as pd
import numpy as np

# Load the dataset
movies = pd.read_csv("movies_dataset.csv")

# Display first few rows
movies.head()
```

```python
# Initial Inspection
movies.shape # Check number of rows and columns
movies.head(10) # Check first 10 rows
movies.tail(10) # Check last 10 rows
```

```python
movies.info() # Basic information about dataset
```

```python
# Summary statistics for numeric columns
movies.describe()
```

```python
# Count missing values in each column
movies.isnull().sum()
```

```python
# Check number of duplicated rows
movies.duplicated().sum()

# Remove duplicate rows if any
movies = movies.drop_duplicates()
```

```python
# Clean column names
movies.columns = movies.columns.str.strip()
movies.columns
```

```python
# Convert ReleaseDate to datetime
movies["ReleaseDate"] = pd.to_datetime(
    movies["ReleaseDate"],
    errors="coerce",
    dayfirst=True
)

movies["ReleaseDate"].head()
```

```python
# Check whether ReleaseYear matches ReleaseDate
movies["Year_from_Date"] = movies["ReleaseDate"].dt.year
movies[["ReleaseYear", "ReleaseDate", "Year_from_Date"]].head()
```

```python
# Check rows where the two years do not match:
year_mismatch = movies[movies["ReleaseYear"] != movies["Year_from_Date"]]
year_mismatch[["Title", "ReleaseYear", "ReleaseDate", "Year_from_Date"]].head()
```

```python
# Convert important numeric columns
numeric_cols = [
    "BudgetUSD",
    "US_BoxOfficeUSD",
    "Global_BoxOfficeUSD",
    "Opening_Day_SalesUSD",
    "One_Week_SalesUSD",
    "IMDbRating",
    "RottenTomatoesScore",
    "NumVotesIMDb",
    "NumVotesRT"
]

for col in numeric_cols:
    movies[col] = pd.to_numeric(movies[col], errors="coerce")
```

```python
movies.dtypes
```

```python
# Handle missing values
movies.isnull().sum()
```

```python
# Create calculated columns
movies["ProfitUSD"] = movies["Global_BoxOfficeUSD"] - movies["BudgetUSD"]
movies["ProfitUSD"]
```

```python
# Return on Investment, ROI
movies["ROI"] = movies["ProfitUSD"] / movies["BudgetUSD"]
movies["ROI"]

# ROI as a percentage
movies["ROI_Percentage"] = movies["ROI"] * 100
movies["ROI_Percentage"]
```

```python
# Save the cleaned dataset
movies.to_csv("cleaned_movies_dataset.csv", index=False)
```

# Data Analysis Questions

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.float_format", "{:,.2f}".format)
```

Q1: Explore trends in movie production, box office, and ratings over time

```python
# 1.1 Number of movies released over time
movies_per_year = (
    movies.groupby("ReleaseYear")["MovieID"]
    .count()
    .reset_index()
    .rename(columns={"MovieID": "Number_of_Movies"})
)

movies_per_year
```

```python
plt.figure(figsize=(10, 6))
plt.plot(
    movies_per_year["ReleaseYear"],
    movies_per_year["Number_of_Movies"],
    marker="o"
)

plt.title("Number of Movies Released Over Time")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.grid(True)
plt.show()
```

- Finding: The analysis shows a clear upward trend in the number of movies released over time. In 1950, the dataset contains around 2,432 movies, while by 2025, the number increases to around 23,809 movies. This suggests strong growth in movie production across the years.

```python
#1.2 Total global box office revenue over time
total_revenue_per_year = (
    movies.groupby("ReleaseYear")["Global_BoxOfficeUSD"]
    .sum()
    .reset_index()
    .rename(columns={"Global_BoxOfficeUSD": "Total_Global_Revenue"})
)

total_revenue_per_year
```

```python
plt.figure(figsize=(10, 6))
plt.plot(
    total_revenue_per_year["ReleaseYear"],
    total_revenue_per_year["Total_Global_Revenue"],
    marker="o"
)

plt.title("Total Global Box Office Revenue Over Time")
plt.xlabel("Release Year")
plt.ylabel("Total Global Box Office Revenue USD")
plt.grid(True)
plt.show()
```

```python
# 1.3 Average global box office revenue over time
avg_revenue_per_year = (
    movies.groupby("ReleaseYear")["Global_BoxOfficeUSD"]
    .mean()
    .reset_index()
    .rename(columns={"Global_BoxOfficeUSD": "Average_Global_Revenue"})
)

avg_revenue_per_year
```

```python
plt.figure(figsize=(11, 5))
plt.plot(
    avg_revenue_per_year["ReleaseYear"],
    avg_revenue_per_year["Average_Global_Revenue"],
    marker="o"
)

plt.title("Average Global Box Office Revenue Over Time")
plt.xlabel("Release Year")
plt.ylabel("Average Global Box Office Revenue USD")
plt.grid(True)
plt.show()
```

- Total global box office revenue also shows a strong increasing trend. Revenue rises from around $72 billion in 1950 to more than $647 billion by 2025. This increase is likely connected to the rising number of movies released each year. As more movies are produced, total box office revenue also increases.

- However, the trend for average global box office revenue per movie is different. Unlike total revenue, average revenue does not show a strong upward trend. It fluctuates over time, mostly staying around $25 million to $30 million per movie. This means that although the movie industry produced more films and generated higher total revenue, the average revenue earned by each individual movie remained relatively stable.

```python
# 1.4 Average IMDb rating over time
avg_imdb_rating_per_year = (
    movies.groupby("ReleaseYear")["IMDbRating"]
    .mean()
    .reset_index()
    .rename(columns={"IMDbRating": "Average_IMDb_Rating"})
)

avg_imdb_rating_per_year.head()
```

```python
plt.figure(figsize=(11, 5))
plt.plot(
    avg_imdb_rating_per_year["ReleaseYear"],
    avg_imdb_rating_per_year["Average_IMDb_Rating"],
    marker="o"
)

plt.title("Average IMDb Rating Over Time")
plt.xlabel("Release Year")
plt.ylabel("Average IMDb Rating")
plt.grid(True)
plt.show()
```

```python
# 1.5 Average Rotten Tomatoes score over time
avg_rt_score_per_year = (
    movies.groupby("ReleaseYear")["RottenTomatoesScore"]
    .mean()
    .reset_index()
    .rename(columns={"RottenTomatoesScore": "Average_RottenTomatoes_Score"})
)

avg_rt_score_per_year.head()
```

```python
plt.figure(figsize=(11, 5))
plt.plot(
    avg_rt_score_per_year["ReleaseYear"],
    avg_rt_score_per_year["Average_RottenTomatoes_Score"],
    marker="o"
)

plt.title("Average Rotten Tomatoes Score Over Time")
plt.xlabel("Release Year")
plt.ylabel("Average Rotten Tomatoes Score")
plt.grid(True)
plt.show()
```

- The average IMDb rating also remains stable over time. Most yearly averages are around 6.5, with only small fluctuations. This suggests that the overall audience rating quality of movies did not change dramatically across the years.

- Similarly, the average Rotten Tomatoes score stays around 64 to 65. There are some noticeable dips and peaks, but the overall pattern is quite stable. This indicates that critic scores also remained relatively consistent over time.

```python
# 1.6 Combined yearly summary table
yearly_summary = (
    movies.groupby("ReleaseYear")
    .agg(
        Number_of_Movies=("MovieID", "count"),
        Total_Global_Revenue=("Global_BoxOfficeUSD", "sum"),
        Average_Global_Revenue=("Global_BoxOfficeUSD", "mean"),
        Average_Budget=("BudgetUSD", "mean"),
        Average_IMDb_Rating=("IMDbRating", "mean"),
        Average_RottenTomatoes_Score=("RottenTomatoesScore", "mean")
    )
    .reset_index()
)

yearly_summary
```

Summary Findings from Q1:
- The yearly trend analysis shows that movie production increased steadily from 1950 to 2025. The number of movies released rose significantly over time, indicating strong growth in the movie industry. 
- Total global box office revenue also increased substantially, which is likely influenced by the higher number of movies released each year. However, average global revenue per movie remained relatively stable, fluctuating mostly between approximately $25 million and $30 million. This suggests that the growth in total revenue was mainly driven by higher movie production volume rather than a major increase in average revenue per film. 
- Average IMDb ratings stayed close to 6.5, while Rotten Tomatoes scores remained around 64 to 65, showing that movie ratings were relatively stable over time. 
- Overall, the industry expanded in terms of production and total revenue, but average movie-level performance and ratings remained fairly consistent.

Q2: What is the relationship between a movie's production budget and its financial success? 
- Investigate both total revenue and profitability. 
- Is a higher budget a guarantee of a higher return on investment?
- Budget vs. Revenue: What is the relationship between a movie's budget and its worldwide revenue? 
- Are higher-budget films guaranteed to be more profitable?

To examine the relationship between production budget and financial success, two calculated variables were created: profit and return on investment. Profit was calculated as global box office revenue minus production budget, while ROI was calculated as profit divided by budget and multiplied by 100. Correlation coefficients were also calculated to measure the strength and direction of these relationships.

```python
# Handle possible infinite values
movies.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
# Check the relevant columns
movies[["Title", "BudgetUSD", "Global_BoxOfficeUSD", "ProfitUSD", "ROI_Percentage"]].head()
```

```python
# Calculate correlation between budget and global revenue
budget_revenue_corr = movies["BudgetUSD"].corr(movies["Global_BoxOfficeUSD"])
budget_revenue_corr
```

```python
# Scatter plot of budget vs global revenue
plt.figure(figsize=(10, 6))

plt.scatter(
    movies["BudgetUSD"],
    movies["Global_BoxOfficeUSD"],
    alpha=0.5
)

plt.title("Budget vs Global Box Office Revenue")
plt.xlabel("Budget USD")
plt.ylabel("Global Box Office Revenue USD")
plt.grid(True)
plt.show()
```

```python
# Calculate correlation between budget and profit
budget_profit_corr = movies["BudgetUSD"].corr(movies["ProfitUSD"])
budget_profit_corr
```

```python
# Scatter plot of budget vs profit
plt.figure(figsize=(10, 6))

plt.scatter(
    movies["BudgetUSD"],
    movies["ProfitUSD"],
    alpha=0.5
)

plt.axhline(0, linestyle="--")

plt.title("Budget vs Profit")
plt.xlabel("Budget USD")
plt.ylabel("Profit USD")
plt.grid(True)
plt.show()
```

```python
# Calculate correlation between budget and ROI
budget_roi_corr = movies["BudgetUSD"].corr(movies["ROI_Percentage"])
budget_roi_corr
```

```python
# Scatter plot of budget vs ROI
plt.figure(figsize=(10, 6))

plt.scatter(
    movies["BudgetUSD"],
    movies["ROI_Percentage"],
    alpha=0.5
)

plt.title("Budget vs Return on Investment ROI")
plt.xlabel("Budget USD")
plt.ylabel("ROI Percentage")
plt.grid(True)
plt.show()
```

```python
# Create a correlation summary table
budget_correlation_summary = pd.DataFrame({
    "Relationship": [
        "Budget vs Global Revenue",
        "Budget vs Profit",
        "Budget vs ROI"
    ],
    "Correlation_Coefficient": [
        budget_revenue_corr,
        budget_profit_corr,
        budget_roi_corr
    ]
})

budget_correlation_summary
```

```python
# Top 10 movies by profit
top_10_profit = (
    movies[["Title", "BudgetUSD", "Global_BoxOfficeUSD", "ProfitUSD", "ROI_Percentage"]]
    .sort_values(by="ProfitUSD", ascending=False)
    .head(10)
)

top_10_profit
```

```python
# Top 10 movies by ROI
top_10_roi = (
    movies[["Title", "BudgetUSD", "Global_BoxOfficeUSD", "ProfitUSD", "ROI_Percentage"]]
    .sort_values(by="ROI_Percentage", ascending=False)
    .head(10)
)

top_10_roi
```

```python
# Filter very small budgets for ROI analysis
top_10_roi_filtered = (
    movies[movies["BudgetUSD"] >= 100000]
    [["Title", "BudgetUSD", "Global_BoxOfficeUSD", "ProfitUSD", "ROI_Percentage"]]
    .sort_values(by="ROI_Percentage", ascending=False)
    .head(10)
)

top_10_roi_filtered
```

Findings from Q2:
- The analysis shows that production budget is strongly associated with global box office revenue. The correlation between budget and global revenue is 0.90, indicating that higher-budget movies generally generate higher worldwide revenue. 
- Budget also has a strong positive relationship with profit, with a correlation of 0.79. This suggests that larger production budgets are often linked to higher absolute profits. 
- However, the relationship between budget and ROI is almost zero, with a correlation of -0.001. This means that a higher budget does not guarantee a higher return on investment. While high-budget movies may earn more total revenue and profit, ROI depends on how much profit is generated relative to the original budget. Therefore, lower-budget movies can sometimes achieve equally strong or even better investment returns. 
- Overall, budget appears to be important for revenue and profit, but it is not a reliable predictor of ROI.

Q3: Which movie genres are the most popular and profitable over time?

```python
# Create a genre-level DataFrame
movies_genre = movies.copy()

movies_genre["Genre"] = movies_genre["Genre"].astype(str)
movies_genre["Genre"] = movies_genre["Genre"].str.split("|")
movies_genre = movies_genre.explode("Genre")
movies_genre["Genre"] = movies_genre["Genre"].str.strip()

movies_genre.head()
```

```python
# Check the cleaned genre column
movies_genre["Genre"].value_counts()
```

```python
# Most popular genres by number of movies
genre_count = (
    movies_genre.groupby("Genre")["MovieID"]
    .count()
    .reset_index()
    .rename(columns={"MovieID": "Number_of_Movies"})
    .sort_values(by="Number_of_Movies", ascending=False)
)

genre_count
```

```python
# Visualize Most popular genres by number of movies
plt.figure(figsize=(12, 6))

plt.bar(
    genre_count["Genre"],
    genre_count["Number_of_Movies"]
)

plt.title("Number of Movies by Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
# Genre performance summary
genre_summary = (
    movies_genre.groupby("Genre")
    .agg(
        Number_of_Movies=("MovieID", "count"),
        Total_Global_Revenue=("Global_BoxOfficeUSD", "sum"),
        Average_Global_Revenue=("Global_BoxOfficeUSD", "mean"),
        Total_Profit=("ProfitUSD", "sum"),
        Average_Profit=("ProfitUSD", "mean"),
        Average_ROI=("ROI_Percentage", "mean"),
        Average_IMDb_Rating=("IMDbRating", "mean"),
        Average_RottenTomatoes_Score=("RottenTomatoesScore", "mean")
    )
    .reset_index()
    .sort_values(by="Total_Profit", ascending=False)
)

genre_summary
```

```python
# Top genres by total profit
top_genres_profit = genre_summary.head(10)
top_genres_profit
```

```python
# Visualize Top 10 Genres by Total Profit
plt.figure(figsize=(12, 6))

plt.bar(
    top_genres_profit["Genre"],
    top_genres_profit["Total_Profit"]
)

plt.title("Top 10 Genres by Total Profit")
plt.xlabel("Genre")
plt.ylabel("Total Profit USD")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
# Top genres by average ROI
genre_roi = genre_summary[genre_summary["Number_of_Movies"] >= 5]

genre_roi = genre_roi.sort_values(by="Average_ROI", ascending=False)
genre_roi.head(10)
```

```python
# Visualize Top 10 Genres by Average ROI
top_genres_roi = genre_roi.head(10)

plt.figure(figsize=(12, 6))

plt.bar(
    top_genres_roi["Genre"],
    top_genres_roi["Average_ROI"]
)

plt.title("Top 10 Genres by Average ROI")
plt.xlabel("Genre")
plt.ylabel("Average ROI Percentage")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
# Genre Trends Over Time
genre_year_trend = (
    movies_genre.groupby(["ReleaseYear", "Genre"])["MovieID"]
    .count()
    .reset_index()
    .rename(columns={"MovieID": "Number_of_Movies"})
)

genre_year_trend.head(10)
```

```python
# Select the top 5 most common genres
top_5_genres = genre_count.head(5)["Genre"].tolist()

# Filter the trend data for top 5 genres
genre_year_top5 = genre_year_trend[
    genre_year_trend["Genre"].isin(top_5_genres)
]

genre_year_top5.head()

```

```python
# Visualize Genre popularity over time
plt.figure(figsize=(12, 6))

for genre in top_5_genres:
    genre_data = genre_year_top5[
        genre_year_top5["Genre"] == genre
    ]

    plt.plot(
        genre_data["ReleaseYear"],
        genre_data["Number_of_Movies"],
        marker="o",
        label=genre
    )

plt.title("Movie Production Trends of Top 5 Genres Over Time")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Genre profitability over time
genre_profit_year = (
    movies_genre.groupby(["ReleaseYear", "Genre"])["ProfitUSD"]
    .sum()
    .reset_index()
    .rename(columns={"ProfitUSD": "Total_Profit"})
)

genre_profit_year.head()
```

```python
# Filter profit trend for top 5 genres
genre_profit_year_top5 = genre_profit_year[
    genre_profit_year["Genre"].isin(top_5_genres)
]

genre_profit_year_top5.head()
```

```python
# Visualize Genre profitability over time
plt.figure(figsize=(12, 6))

for genre in top_5_genres:
    genre_data = genre_profit_year_top5[
        genre_profit_year_top5["Genre"] == genre
    ]

    plt.plot(
        genre_data["ReleaseYear"],
        genre_data["Total_Profit"],
        marker="o",
        label=genre
    )

plt.title("Profit Trends of Top 5 Genres Over Time")
plt.xlabel("Release Year")
plt.ylabel("Total Profit USD")
plt.legend()
plt.grid(True)
plt.show()
```

Findings from Q3:
- To analyze genre performance, the genre column was first split so that movies with multiple genres could be counted under each relevant genre. The results show that Drama is the most frequently produced genre, followed by Comedy and Action. Drama also generated the highest total global revenue and total profit, suggesting that it is the strongest genre in terms of overall commercial contribution. However, this result is partly influenced by the large number of Drama movies in the dataset.
- When looking at average profitability, the differences between genres are smaller. Most genres generated average profits of around $17 million to $18 million per movie. Horror and Documentary showed slightly higher average profit compared with several other genres. 
- In terms of ROI, Horror had the highest average ROI, followed by Action, although ROI values were relatively similar across genres. This suggests that while Drama dominates in total revenue and total profit, Horror may be slightly more efficient in generating returns relative to budget.
- Over time, the production of the top genres increased steadily. Drama remained the most produced genre throughout the period, followed by Comedy and Action. The profitability trend also increased across all top genres, with Drama consistently leading in total profit. 
- Overall, the findings suggest that Drama, Comedy, and Action dominate the industry in total scale, while Horror performs slightly better when profitability is evaluated relative to investment.

Q4: The "Best" Directors: Which directors consistently deliver high-rated and commercially successful films? 

```python
# Check the Director column
movies["Director"].head()
```

```python
movies["Director"].isnull().sum()
```

```python
movies["Director"].nunique()
```

```python
# Make sure Profit and ROI columns exist
movies["ProfitUSD"] = movies["Global_BoxOfficeUSD"] - movies["BudgetUSD"]

movies["ROI_Percentage"] = (
    movies["ProfitUSD"] / movies["BudgetUSD"]
) * 100

movies.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
# Group by Director
director_summary = (
    movies.groupby("Director")
    .agg(
        Number_of_Movies=("MovieID", "count"),
        Average_IMDb_Rating=("IMDbRating", "mean"),
        Average_RottenTomatoes_Score=("RottenTomatoesScore", "mean"),
        Total_Global_Revenue=("Global_BoxOfficeUSD", "sum"),
        Average_Global_Revenue=("Global_BoxOfficeUSD", "mean"),
        Total_Profit=("ProfitUSD", "sum"),
        Average_Profit=("ProfitUSD", "mean"),
        Average_ROI=("ROI_Percentage", "mean")
    )
    .reset_index()
)

director_summary.head()
```

```python
# Filter directors with more than 3 movies
director_summary_filtered = director_summary[
    director_summary["Number_of_Movies"] > 3
]

director_summary_filtered.head()
```

```python
# Check how many directors remain
director_summary_filtered.shape
```

```python
# Best directors by average IMDb rating
best_directors_rating = (
    director_summary_filtered
    .sort_values(by="Average_IMDb_Rating", ascending=False)
    .head(10)
)

best_directors_rating
```

```python
# Visualize Top directors by IMDb rating

best_directors_rating_plot = best_directors_rating.sort_values(
    by="Average_IMDb_Rating",
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    best_directors_rating_plot["Director"],
    best_directors_rating_plot["Average_IMDb_Rating"]
)

plt.title("Top 10 Directors by Average IMDb Rating")
plt.xlabel("Average IMDb Rating")
plt.ylabel("Director")

plt.xlim(
    best_directors_rating_plot["Average_IMDb_Rating"].min() - 0.01,
    best_directors_rating_plot["Average_IMDb_Rating"].max() + 0.01
)

plt.grid(axis="x")
plt.show()
```

```python
# Best directors by Rotten Tomatoes score
best_directors_rt = (
    director_summary_filtered
    .sort_values(by="Average_RottenTomatoes_Score", ascending=False)
    .head(10)
)

best_directors_rt
```

```python
# Visualize Top directors by Rotten Tomatoes score
best_directors_rt_plot = best_directors_rt.sort_values(
    by="Average_RottenTomatoes_Score",
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    best_directors_rt_plot["Director"],
    best_directors_rt_plot["Average_RottenTomatoes_Score"]
)

plt.title("Top 10 Directors by Average Rotten Tomatoes Score")
plt.xlabel("Average Rotten Tomatoes Score")
plt.ylabel("Director")

plt.xlim(
    best_directors_rt_plot["Average_RottenTomatoes_Score"].min() - 0.1,
    best_directors_rt_plot["Average_RottenTomatoes_Score"].max() + 0.1
)

plt.grid(axis="x")
plt.show()
```

```python
# visualize Top directors by average ROI
best_directors_roi_plot = best_directors_roi.sort_values(
    by="Average_ROI",
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    best_directors_roi_plot["Director"],
    best_directors_roi_plot["Average_ROI"]
)

plt.title("Top 10 Directors by Average ROI")
plt.xlabel("Average ROI Percentage")
plt.ylabel("Director")

plt.xlim(
    best_directors_roi_plot["Average_ROI"].min() - 1,
    best_directors_roi_plot["Average_ROI"].max() + 1
)

plt.grid(axis="x")
plt.show()
```

```python
# Overall ranking of directors
director_rank = director_summary_filtered.copy()

director_rank["IMDb_Rank"] = director_rank["Average_IMDb_Rating"].rank(ascending=False)
director_rank["RT_Rank"] = director_rank["Average_RottenTomatoes_Score"].rank(ascending=False)
director_rank["Profit_Rank"] = director_rank["Total_Profit"].rank(ascending=False)
director_rank["ROI_Rank"] = director_rank["Average_ROI"].rank(ascending=False)

director_rank["Overall_Rank_Score"] = (
    director_rank["IMDb_Rank"] +
    director_rank["RT_Rank"] +
    director_rank["Profit_Rank"] +
    director_rank["ROI_Rank"]
)

best_directors_overall = (
    director_rank
    .sort_values(by="Overall_Rank_Score", ascending=True)
    .head(10)
)

best_directors_overall
```

```python
# Visualize Top directors by overall rank score
plt.figure(figsize=(12, 6))

plt.bar(
    best_directors_overall["Director"],
    best_directors_overall["Overall_Rank_Score"]
)

plt.title("Top 10 Overall Best Directors")
plt.xlabel("Director")
plt.ylabel("Overall Rank Score")
plt.xticks(rotation=45, ha="right")
plt.show()
```
